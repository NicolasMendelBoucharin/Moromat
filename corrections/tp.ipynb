## Exercice 2.

### a.

$(E_1)$ donne $y(0)=1$ donc dans le schmé d'Euler correspondant $y_0=1$. Par ailleurs on a $y'(t)=y(t)$ d'où $y'(t)=f(y(t))$ avec $f(t)=t$. Donc dans le schéma d'Euler correspondant $y_{n+1} = y_n + h*f(y_n)=y_n + hy_n$


### b.

Je vais directement le faire en python.

### c.

Plutôt que de refaire un algorithme à chaque question je vais faire un algorithme générique qui mange la fonction $f$, un point de depart $y_0$ un intervalle $[a,b]$ et un nombre de subdivisions $N$.


```python
def euler(N, a, b, f, y0):
    h=(b-a)/N
    Lt=[a + i*h for i in range(N+1)]
    y=[y0]
    for i in range(1, N+1):
        yi=y[i-1]+h*f(y[i-1])
        y.append(yi)
    plt.plot(Lt, y, label=f"Pour n={N}")
    return y
```

Et pour notre cas précis :

```python
def f(x):
    return x

def vrai_solution(x):
    return np.exp(x)
    
a=0
b=1
y0=1

plt.clf() #On nettoie au cas où
for N in [5, 10, 20, 50]: #on fait tout les cas de N possibles
    euler(N, a,b,f,y0)#on fait euler pour tout les N


#On fait maintenant une autre subdivision pour afficher la vrai solution
N=50
h=h=(b-a)/N
Lt_vrai=[a + i*h for i in range(N+1)]
plt.plot(Lt_vrai, [vrai_solution(x) for x in Lt_vrai], label="e^x")

plt.legend()#Affiche des legendes
plt.show()#affiche le graphe

```
